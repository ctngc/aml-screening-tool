```python
import pandas as pd
df = pd.read_excel("clients_fictifs_AML.xlsx")
df

```

```python
df_sanctions = pd.read_excel("liste_sanctions_fictive.xlsx")
df_sanctions
```

```python
matches_exacts = df[df["Nom_Complet"].isin(df_sanctions["Nom_Liste"])]
matches_exacts
```

```python
!pip install rapidfuzz
```

```python
from rapidfuzz import fuzz

resultats = []

for nom_client in df["Nom_Complet"]:
    for nom_sanction in df_sanctions["Nom_Liste"]:
        score = fuzz.ratio(nom_client, nom_sanction)
        if score > 80:  # seuil de similarité à ajuster
            resultats.append({
                "Client": nom_client,
                "Nom_Sanction_Proche": nom_sanction,
                "Score_Similarite": score
            })

alertes = pd.DataFrame(resultats)
alertes.sort_values("Score_Similarite", ascending=False)
```

```python
from datetime import datetime

with pd.ExcelWriter("rapport_AML.xlsx", engine="openpyxl") as writer:

    # Feuille 1 : correspondances exactes sur la liste de sanctions
    matches_exacts.to_excel(writer, sheet_name="Sanctions_Exactes", index=False)

    # Feuille 2 : correspondances approximatives (fuzzy matching)
    alertes.to_excel(writer, sheet_name="Sanctions_Fuzzy", index=False)

    # Feuille 3 : transactions suspectes (structuring)
    structuring = df_tx[(df_tx["Montant_CHF"] >= 9000) & (df_tx["Montant_CHF"] <= 9999)]
    structuring.to_excel(writer, sheet_name="Structuring", index=False)

    # Feuille 4 : transactions à montant inhabituel
    gros_montants = df_tx[df_tx["Montant_CHF"] > 100000]
    gros_montants.to_excel(writer, sheet_name="Montants_Eleves", index=False)

    # Feuille 5 : résumé synthétique
    resume = pd.DataFrame({
        "Indicateur": [
            "Date du rapport",
            "Nb clients screenés",
            "Nb correspondances exactes",
            "Nb correspondances fuzzy",
            "Nb transactions structuring",
            "Nb transactions montant élevé"
        ],
        "Valeur": [
            datetime.now().strftime("%Y-%m-%d"),
            len(df),
            len(matches_exacts),
            len(alertes),
            len(structuring),
            len(gros_montants)
        ]
    })
    resume.to_excel(writer, sheet_name="Resume", index=False)

print("Rapport généré avec succès")
```
